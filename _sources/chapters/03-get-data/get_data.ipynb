```python
import cdsapi
import zipfile
import os
```

```python
dataset = "derived-drought-historical-monthly"
cds_kwargs = {
    'url': 'https://cds.climate.copernicus.eu/api',
    'key': None
}

client = cdsapi.Client(**cds_kwargs)
years = [str(y) for y in range(1940, 2017)]
months = [f"{m:02d}" for m in range(1, 13)]
accumulation_periods = ["1", "3", "6", "12", "24", "36", "48"]

for year in years:
    for month in months:
        print(f"Processing {year}-{month}...")
        # for acc in accumulation_periods:
        request = {
            "variable": [
                # "probability_of_zero_precipitation_spi",
                # "test_for_normality_spi",
                # "test_for_normality_spei"
                "standardised_precipitation_index",
                "standardised_precipitation_evapotranspiration_index",
            ],
            "accumulation_period": ["1", "3", "6", "12", "24", "36", "48"],
            "version": "1_0",
            "product_type": ["reanalysis"],
            "dataset_type": "consolidated_dataset",
            "year": [year],
            "month": [month],
        }

        try:
            target = f"{dataset}_{year}_{month}.zip"
            if os.path.exists(target):
                with zipfile.ZipFile(target, 'r') as zip_ref:
                    zip_ref.extractall("data/")
                    os.remove(target)
            elif os.path.exists(f"data/SPEI1_genlogistic_global_era5_moda_ref1991to2020_{year}{month}.nc"):
                continue
            else: 
                print(f"Retrieving {year}-{month}...")
                client.retrieve(dataset, request).download(target)
                print(f"Downloaded: {target}")
                with zipfile.ZipFile(target, 'r') as zip_ref:
                    zip_ref.extractall("data/")
                    os.remove(target)
        except Exception as e:
            print(f"Skipped {year}-{month}: {e}")


```
